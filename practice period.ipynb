Training

```python
features = [
    "cycle_length_days",
    "prev_cycle_length",
    "flow_level",
    "pain_level",
    "pms_symptoms",
    "mood_score",
    "stress_score_cycle",
    "sleep_hours_cycle",
    "energy_level",
    "concentration_score",
    "work_hours_lost",
    "prepared_before_period"
]

target = "ovulation_result"

X = data[features].copy()
y = data[target].copy()

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget:")
print(y.value_counts())
```

Identify Categorical Columns

```python
categorical_features = [
    "flow_level",
    "pms_symptoms"
]

numerical_features = [
    "cycle_length_days",
    "prev_cycle_length",
    "pain_level",
    "mood_score",
    "stress_score_cycle",
    "sleep_hours_cycle",
    "energy_level",
    "concentration_score",
    "work_hours_lost",
    "prepared_before_period"
]

print("Categorical:", categorical_features)
print("Numerical:", numerical_features)
```

Train / Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nTraining target:")
print(y_train.value_counts())

print("\nTesting target:")
print(y_test.value_counts())
```

Processing Pipeline

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numerical_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print("Preprocessing pipeline created successfully.")
```

Random Forest Model

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        class_weight="balanced"
    ))
])

rf_model.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

Prediction

```python
y_pred = rf_model.predict(X_test)

print("Predictions completed!")
print(y_pred[:20])
```

Evaluation

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

Multiple models

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    
    "Decision Tree": DecisionTreeClassifier(
        random_state=42,
        max_depth=8
    ),
    
    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        class_weight="balanced"
    ),
    
    "KNN": KNeighborsClassifier(
        n_neighbors=7
    ),
    
    "SVM": SVC(
        kernel="rbf",
        random_state=42
    ),
    
    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=100,
        random_state=42
    )
}
```

Model Train

```python
from sklearn.metrics import accuracy_score, f1_score

results = []

for name, model in models.items():

    pipeline = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("classifier", model)
    ])

    pipeline.fit(X_train, y_train)

    prediction = pipeline.predict(X_test)

    accuracy = accuracy_score(y_test, prediction)
    f1 = f1_score(y_test, prediction, pos_label="Positive")

    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "F1_Score": f1
    })

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="Accuracy",
    ascending=False
)

print(results_df)
```

```python
features_v2 = [
    "cycle_length_days",
    "prev_cycle_length",
    "flow_level",
    "pain_level",
    "pms_symptoms",
    "mood_score",
    "stress_score_cycle",
    "sleep_hours_cycle",
    "energy_level",
    "concentration_score",
    "work_hours_lost",
    "estrogen_pgml",
    "progesterone_ngml",
    "overall_health_score",
    "prepared_before_period"
]

X2 = data[features_v2].copy()
y2 = data["ovulation_result"].copy()

print(X2.shape)
print(X2.head())
```

V2 preprocessing

```python
categorical_features_v2 = [
    "flow_level",
    "pms_symptoms"
]

numerical_features_v2 = [
    "cycle_length_days",
    "prev_cycle_length",
    "pain_level",
    "mood_score",
    "stress_score_cycle",
    "sleep_hours_cycle",
    "energy_level",
    "concentration_score",
    "work_hours_lost",
    "estrogen_pgml",
    "progesterone_ngml",
    "overall_health_score",
    "prepared_before_period"
]

preprocessor_v2 = ColumnTransformer(
    transformers=[
        ("num",
         SimpleImputer(strategy="median"),
         numerical_features_v2),

        ("cat",
         Pipeline([
             ("imputer", SimpleImputer(strategy="most_frequent")),
             ("encoder", OneHotEncoder(handle_unknown="ignore"))
         ]),
         categorical_features_v2)
    ]
)

print("V2 preprocessing ready!")
```

Train/Test Split

```python
X2_train, X2_test, y2_train, y2_test = train_test_split(
    X2,
    y2,
    test_size=0.20,
    random_state=42,
    stratify=y2
)

print("Training:", X2_train.shape)
print("Testing:", X2_test.shape)
```

Compare models Again

```python
models_v2 = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    
    "Decision Tree": DecisionTreeClassifier(
        random_state=42,
        max_depth=8
    ),
    
    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        class_weight="balanced"
    ),
    
    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=100,
        random_state=42
    )
}
```

```python
results_v2 = []

for name, model in models_v2.items():

    pipeline = Pipeline([
        ("preprocessor", preprocessor_v2),
        ("classifier", model)
    ])

    pipeline.fit(X2_train, y2_train)

    pred = pipeline.predict(X2_test)

    acc = accuracy_score(y2_test, pred)
    f1 = f1_score(y2_test, pred, pos_label="Positive")

    results_v2.append({
        "Model": name,
        "Accuracy": acc,
        "F1_Score": f1
    })

results_v2_df = pd.DataFrame(results_v2)

results_v2_df = results_v2_df.sort_values(
    "Accuracy",
    ascending=False
)

print(results_v2_df)
```

```python
from sklearn.ensemble import GradientBoostingClassifier

gb_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", GradientBoostingClassifier(random_state=42))
])

gb_pipeline.fit(X2_train, y2_train)

y2_pred = gb_pipeline.predict(X2_test)

print("Accuracy:", accuracy_score(y2_test, y2_pred))

print("\nClassification Report:")
print(classification_report(y2_test, y2_pred))
```

```python
def predict_ovulation():
    cycle_length = float(input("Cycle length (days): "))
    prev_cycle = float(input("Previous cycle length (days): "))
    flow = input("Flow level (Light/Medium/Heavy): ")
    pain = float(input("Pain level (0-10): "))
    pms = input("PMS symptoms (None/Mild/Moderate/Severe): ")
    mood = float(input("Mood score (0-10): "))
    stress = float(input("Stress score (0-10): "))
    sleep = float(input("Sleep hours: "))
    energy = float(input("Energy level (0-10): "))
    concentration = float(input("Concentration score (0-10): "))
    work_lost = float(input("Work/study hours lost: "))
    estrogen = float(input("Estrogen level (pg/mL): "))
    progesterone = float(input("Progesterone level (ng/mL): "))
    health_score = float(input("Overall health score: "))
    prepared = int(input("Prepared before period? (1=Yes, 0=No): "))

    user_data = pd.DataFrame({
        "cycle_length_days": [cycle_length],
        "prev_cycle_length": [prev_cycle],
        "flow_level": [flow],
        "pain_level": [pain],
        "pms_symptoms": [pms],           #premenstrual Syndrome
        "mood_score": [mood],
        "stress_score_cycle": [stress],
        "sleep_hours_cycle": [sleep],
        "energy_level": [energy],
        "concentration_score": [concentration],
        "work_hours_lost": [work_lost],
        "estrogen_pgml": [estrogen],
        "progesterone_ngml": [progesterone],
        "overall_health_score": [health_score],
        "prepared_before_period": [prepared]
    })

    prediction = gb_pipeline.predict(user_data)[0]
    probability = gb_pipeline.predict_proba(user_data)[0]

    print("\n========== HerHealth AI ==========")
    print("Predicted Ovulation Result:", prediction)
    print("Negative Probability:", round(probability[0] * 100, 2), "%")
    print("Positive Probability:", round(probability[1] * 100, 2), "%")

    print("\nNote: This is an AI-based screening prediction,")
    print("not a medical diagnosis.")
```
